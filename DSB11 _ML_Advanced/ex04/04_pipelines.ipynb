## 1. Preprocessing pipeline

Create three custom transformers, the first two out of which will be used within a [Pipeline](https://scikit-learn.org/stable/modules/generated/sklearn.pipeline.Pipeline.html).

1. `FeatureExtractor()` class:
 - Takes a dataframe with `uid`, `labname`, `numTrials`, `timestamp` from the file [`checker_submits.csv`](https://drive.google.com/file/d/14voc4fNJZiLEFaZyd8nEG-lQt5JjatYw/view?usp=sharing).
 - Extracts `hour` from `timestamp`.
 - Extracts `weekday` from `timestamp` (numbers).
 - Drops the `timestamp` column.
 - Returns the new dataframe.


2. `MyOneHotEncoder()` class:
 - Takes the dataframe from the result of the previous transformation and the name of the target column.
 - Identifies all the categorical features and transforms them with `OneHotEncoder()`. If the target column is categorical too, then the transformation should not apply to it.
 - Drops the initial categorical features.
 - Returns the dataframe with the features and the series with the target column.


3. `TrainValidationTest()` class:
 - Takes `X` and `y`.
 - Returns `X_train`, `X_valid`, `X_test`, `y_train`, `y_valid`, `y_test` (`test_size=0.2`, `random_state=21`, `stratified`).


````python
class FeatureExtractor(BaseEstimator, TransformerMixin):
    def __init__(self):
        pass

    def fit(self, X, y=None):
        return self

    def transform(self, X, y=None):
        df = X.copy()

      
        if 'timestamp' not in df.columns:
            raise ValueError("В DataFrame отсутствует колонка 'timestamp'.")

        if not pd.api.types.is_datetime64_any_dtype(df['timestamp']):
            df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')


        df['hour'] = df['timestamp'].dt.hour
        df['dayOfWeek'] = df['timestamp'].dt.day_of_week


        df = df.drop(columns=['timestamp'])

        return df
````

````python
class MyOneHotEncoder(BaseEstimator, TransformerMixin):
    def __init__(self, target_column):
        self.target_column = target_column
        self.encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
        self.scaler = StandardScaler()

    def fit(self, X, y=None):
        self.categorical_cols = ['uid', 'labname']
        self.numeric_cols = ['numTrials', 'hour']

        self.encoder.fit(X[self.categorical_cols])
        self.scaler.fit(X[self.numeric_cols])
        return self

    def transform(self, X, y=None):
        encoded_data = self.encoder.transform(X[self.categorical_cols])
        encoded_df = pd.DataFrame(
            encoded_data,
            columns=self.encoder.get_feature_names_out(self.categorical_cols),
            index=X.index
        )

        scaled_data = self.scaler.transform(X[self.numeric_cols])
        scaled_df = pd.DataFrame(scaled_data, columns=self.numeric_cols, index=X.index)

        X_transformed = pd.concat([scaled_df, encoded_df], axis=1)

  
        if self.target_column in X.columns:
            X_transformed[self.target_column] = X[self.target_column]

        return X_transformed
````

````python
class TrainValidationTest(BaseEstimator, TransformerMixin):
    def __init__(self, test_size=0.2, valid_size=0.2, random_state=21):
        self.test_size = test_size
        self.valid_size = valid_size
        self.random_state = random_state

    def fit(self, X, y=None):
        return self

    def transform(self, X, y):
        if y is None:
            raise ValueError("Необходимо передать y при вызове transform.")

        X_train_val, X_test, y_train_val, y_test = train_test_split(
            X, y,
            test_size=self.test_size,
            random_state=self.random_state,
            stratify=y
        )

        X_train, X_valid, y_train, y_valid = train_test_split(
            X_train_val, y_train_val,
            test_size=self.valid_size,
            random_state=self.random_state,
            stratify=y_train_val
        )

        return X_train, X_valid, X_test, y_train, y_valid, y_test
````

## 2. Model selection pipeline

````python
class ModelSelection:
    def __init__(self, grids, grids_dict):
        self.grids = grids
        self.grids_dict = grids_dict
        self.results = []

    def choose(self, X_train, y_train, X_valid, y_valid):
        best_valid_score = 0
        best_model_name = None

        for idx, grid in enumerate(self.grids):
            estimator_name = self.grids_dict[idx]
            print(f"\nEstimator: {estimator_name}")

            # Обучение с прогрессом по числу комбинаций параметров
            with tqdm(total=None, desc=f"GridSearchCV for {estimator_name}") as pbar:
                grid.fit(X_train, y_train)
                pbar.update()

            best_params = grid.best_params_
            best_train_score = grid.best_score_

            y_pred_valid = grid.best_estimator_.predict(X_valid)
            valid_score = accuracy_score(y_valid, y_pred_valid)

            print(f"Best params: {best_params}")
            print(f"Best training accuracy: {best_train_score:.3f}")
            print(f"Validation set accuracy score for best params: {valid_score:.3f}")

            self.results.append({
                'model': estimator_name,
                'params': best_params,
                'valid_score': valid_score
            })

            if valid_score > best_valid_score:
                best_valid_score = valid_score
                best_model_name = estimator_name

        print(f"\nClassifier with best validation set accuracy: {best_model_name}")
        return best_model_name

    def best_results(self):
        return pd.DataFrame(self.results)
````

````python
svm = SVC()
tree = DecisionTreeClassifier()
rf = RandomForestClassifier()


svm_params = {
    'kernel': ['linear', 'rbf', 'sigmoid'],
    'C': [0.01, 0.1, 1, 1.5, 5, 10],
    'gamma': ['scale', 'auto'],
    'class_weight': ['balanced', None],
    'random_state': [21],
    'probability': [True]
}

tree_params = {
    'max_depth': np.arange(1, 50),
    'class_weight': [None, 'balanced'],
    'criterion': ['gini', 'entropy'],
    'random_state': [21]
}

rf_params = {
    'n_estimators': [5, 10, 50, 100],
    'max_depth': np.arange(1, 50),
    'class_weight': [None, 'balanced'],
    'criterion': ['gini', 'entropy'],
    'random_state': [21]
}


jobs = -1
gs_svm = GridSearchCV(estimator=svm, param_grid=svm_params, scoring='accuracy', cv=2, n_jobs=jobs)
gs_tree = GridSearchCV(estimator=tree, param_grid=tree_params, scoring='accuracy', cv=2, n_jobs=jobs)
gs_rf = GridSearchCV(estimator=rf, param_grid=rf_params, scoring='accuracy', cv=2, n_jobs=jobs)


grids = [gs_svm, gs_tree, gs_rf]
grids_dict = {0: 'SVM', 1: 'Decision Tree', 2: 'Random Forest'}
````

## 3. Finalization

`Finalize()` class
 - Takes an estimator.
 - Method `final_score()` takes `X_train`, `y_train`, `X_test`, `y_test` and returns the accuracy of the model as in the example below:
```
final.final_score(X_train, y_train, X_test, y_test)
Accuracy of the final model is 0.908284023668639
```
 - Method `save_model()` takes a path, saves the model to this path and prints that the model was successfully saved.

````python
class Finalize:
    
    def __init__(self, model):
        self.model = model

    def final_score(self, X_train, y_train, X_test, y_test):
     
        self.model.fit(X_train, y_train)
        score = self.model.score(X_test, y_test)
        print(f'Accuracy of the final model is: {score:.6f}')
        return score

    def save_model(self, path):

        joblib.dump(self.model, path)
        print(f'Model successfully saved to {path}')
````

````python
df = pd.read_csv('../data/checker_submits.csv', parse_dates=['timestamp'])
````

````python
preprocessing = Pipeline([
    ('feature_extractor', FeatureExtractor()),
    ('onehot_encoder', MyOneHotEncoder('dayOfWeek'))
])
data = preprocessing.fit_transform(df)
data
````

````python
X = data.drop('dayOfWeek', axis=1)
y = data['dayOfWeek']
````

````python
splitter = TrainValidationTest()
X_train, X_valid, X_test, y_train, y_valid, y_test = splitter.transform(X, y)
````

````python
selection = ModelSelection(grids, grids_dict)
selection.choose(X_train, y_train, X_valid, y_valid)
model_df = selection.best_results()
````
